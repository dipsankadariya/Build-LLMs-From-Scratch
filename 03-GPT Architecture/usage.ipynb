```python
# --------------------------------------------------------------------------
# Example Usage
# --------------------------------------------------------------------------

# sample paragraph
text = """
Large language models learn by predicting the next token in a sequence.
During training, the text is first converted into token IDs.
These token IDs are then split into overlapping input and target sequences.
"""

# create the tokenizer
tokenizer = tiktoken.get_encoding("gpt2")

# create the dataset
dataset = GPTDatasetV1(
    txt=text,
    tokenizer=tokenizer,
    max_length=8,
    stride=4
)

print(f"Total training samples: {len(dataset)}\n")

# exmple each sample
for i in range(len(dataset)):
    input_ids, target_ids = dataset[i]

    print(f"Sample {i + 1}")
    print("=" * 60)

    print("Input IDs:")
    print(input_ids.tolist())

    print("\nTarget IDs:")
    print(target_ids.tolist())

    print("\nDecoded Input:")
    print(tokenizer.decode(input_ids.tolist()))

    print("\nDecoded Target:")
    print(tokenizer.decode(target_ids.tolist()))

    print("\n" + "-" * 60 + "\n")
```
