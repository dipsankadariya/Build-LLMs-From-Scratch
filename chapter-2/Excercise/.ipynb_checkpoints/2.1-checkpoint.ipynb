```python
import tiktoken
text="Akwirw ier"
tokenizer= tiktoken.get_encoding("gpt2")
integers= tokenizer.encode(text)
print(integers)
```

```python
tokenizer.decode(integers)
```

```python
#individual mapping of id with tokens
for id in integers:
    print(f"{id}->{repr(tokenizer.decode([id]))}")
```
